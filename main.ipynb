````python
def get_dupes(df_user):
    duplicate_counts = df_user['reviewerID'].value_counts()
    # Lấy những ID xuất hiện > 1 lần
    print(duplicate_counts[duplicate_counts > 1])

    # keep=False sẽ đánh dấu True cho TẤT CẢ các dòng có reviewerID xuất hiện từ 2 lần trở lên
    df_all_dupes = df_user[df_user.duplicated(subset=['reviewerID'], keep=False)]

    # Sau đó bạn có thể sắp xếp lại để xem cho dễ
    df_all_dupes.sort_values(by='reviewerID', inplace=True)

    return df_all_dupes
````

````python
import torch
from utils.configurator import Config
from utils.dataset import RecDataset
from utils.dataloader import TrainDataLoader
from utils.utils import get_model

def load_model(config: Config):
    # Load dataset
    dataset = RecDataset(config)

    # FIX: Trigger inter_num calculation
    print(dataset)  # This calls __str__() which sets inter_num
    # OR manually set it:
    # dataset.inter_num = len(dataset.df)

    # Now this works
    train_data = TrainDataLoader(config, dataset, batch_size=config['train_batch_size'], shuffle=False)

    # Convert list hyperparameters to scalars (pick first value)
    if isinstance(config['dropout_rate'], list):
        config['dropout_rate'] = config['dropout_rate'][0]
    if isinstance(config['n_ui_layers'], list):
        config['n_ui_layers'] = config['n_ui_layers'][0]
    if isinstance(config['image_knn_k'], list):
        config['image_knn_k'] = config['image_knn_k'][0]
    if isinstance(config['text_knn_k'], list):
        config['text_knn_k'] = config['text_knn_k'][0]
    if isinstance(config['reg_weight'], list):
        config['reg_weight'] = config['reg_weight'][0]

    # Now instantiate model
    model = get_model('SMORE')(config, train_data).to(config['device']) 

    return model
````

````python
import json
import torch
import pandas as pd

def get_recommendations_json(model, df_user_map, df_meta_map, user_ids=[0, 1, 2], top_k=10):
    model.eval()
    all_recommendations = {}

    with torch.no_grad():
        for u_id in user_ids:
            # 1. Dự đoán bằng model
            user_tensor = torch.tensor([u_id])
            scores = model.full_sort_predict(user_tensor)
            _, top_items_indices = torch.topk(scores, top_k)
            top_items_list = top_items_indices[0].tolist()

            # 2. Lấy thông tin người dùng (tên + ID gốc)
            # Dựa trên image_d3042b.png: dùng cột userID, reviewerName, reviewerID
            user_row = df_user_map[df_user_map['userID'] == u_id]
            if not user_row.empty:
                user_info = f"{user_row.iloc[0]['reviewerName']} ({user_row.iloc[0]['reviewerID']})"
            else:
                user_info = f"Unknown User (ID: {u_id})"

            # 3. Lấy danh sách sản phẩm dưới dạng JSON của từng dòng
            # Dựa trên image_d30446.png: dùng cột itemID để lọc
            product_details = []
            for i_id in top_items_list:
                product_row = df_meta_map[df_meta_map['itemID'] == i_id]
                if not product_row.empty:
                    # Chuyển dòng dataframe thành dictionary (json object)
                    # to_dict('records')[0] sẽ lấy dòng đầu tiên dưới dạng dict
                    # Chỉ lấy các cột thông tin cơ bản để JSON nhẹ hơn
                    cols_to_keep = ['itemID', 'asin', 'title', 'price', 'imUrl']
                    item_dict = product_row[cols_to_keep].to_dict('records')[0]
                    product_details(item_dict)
                else:
                    product_details.append({"itemID": i_id, "error": "Item metadata not found"})

            # 4. Gộp vào kết quả tổng với key là userID
            all_recommendations[str(u_id)] = {
                "user_info": user_info,
                "recommended_products": product_details
            }

    # Trả về chuỗi JSON định dạng đẹp
    return json.dumps(all_recommendations, indent=4, ensure_ascii=False)
````

````python
def get_recommendations_json(model, df_user_map, df_meta_map, user_ids=[0, 1, 2], top_k=10):
    model.eval()
    all_recommendations = {}

    with torch.no_grad():
        for u_id in user_ids:
            # 1. Dự đoán
            user_tensor = torch.tensor([u_id])
            scores = model.full_sort_predict(user_tensor) 
            
            _, top_items_indices = torch.topk(scores, top_k)
            top_items_list = top_items_indices.flatten().tolist() 

            # 2. Lấy thông tin người dùng
            user_row = df_user_map[df_user_map['userID'] == u_id]
            if not user_row.empty:
                # Xử lý trường hợp reviewerName bị NaN để tránh lỗi JSON
                name = user_row.iloc[0]['reviewerName']
                name = name if pd.notna(name) else "Unknown Customer"
                user_info = f"{name} ({user_row.iloc[0]['reviewerID']})"
            else:
                user_info = f"Unknown User (ID: {u_id})"

            # 3. Lấy danh sách sản phẩm
            product_details = []
            for i_id in top_items_list:
                product_row = df_meta_map[df_meta_map['itemID'] == i_id]
                if not product_row.empty:
                    # Chỉ lấy các cột cần thiết để tránh dính ndarray (gây lỗi JSON)
                    cols_to_keep = ['itemID', 'asin', 'title', 'price', 'imUrl']
                    # Lọc các cột tồn tại trong dataframe để tránh KeyError
                    existing_cols = [c for c in cols_to_keep if c in product_row.columns]
                    
                    item_dict = product_row[existing_cols].to_dict('records')[0]
                    
                    # Fix các giá trị NaN trong dict sản phẩm (JSON không đọc được NaN)
                    item_dict = {k: (v if pd.notna(v) else None) for k, v in item_dict.items()}
                    
                    # SỬA LỖI TẠI ĐÂY: Dùng .append thay vì gọi như hàm
                    product_details.append(item_dict) 
                else:
                    product_details.append({"itemID": i_id, "error": "Item metadata not found"})

            all_recommendations[str(u_id)] = {
                "user_info": user_info,
                "recommended_products": product_details
            }

    return json.dumps(all_recommendations, indent=4, ensure_ascii=False)
````

````python
# --- CÁCH SỬ DỤNG ---
json_result = get_recommendations_json(model, df_user_map, df_meta_map)
print(json_result)
````
