````python
def get_dupes(df_user):
    duplicate_counts = df_user["reviewerID"].value_counts()
    # Lấy những ID xuất hiện > 1 lần
    print(duplicate_counts[duplicate_counts > 1])

    # keep=False sẽ đánh dấu True cho TẤT CẢ các dòng có reviewerID xuất hiện từ 2 lần trở lên
    df_all_dupes = df_user[df_user.duplicated(subset=["reviewerID"], keep=False)]

    # Sau đó bạn có thể sắp xếp lại để xem cho dễ
    df_all_dupes.sort_values(by="reviewerID", inplace=True)

    return df_all_dupes
````

````python
import torch
from utils.configurator import Config
from utils.dataset import RecDataset
from utils.dataloader import TrainDataLoader
from utils.utils import get_model


def load_model(config: Config):
    # Load dataset
    dataset = RecDataset(config)

    # FIX: Trigger inter_num calculation
    print(dataset)  # This calls __str__() which sets inter_num
    # OR manually set it:
    # dataset.inter_num = len(dataset.df)

    # Now this works
    train_data = TrainDataLoader(
        config, dataset, batch_size=config["train_batch_size"], shuffle=False
    )

    # Convert list hyperparameters to scalars (pick first value)
    if isinstance(config["dropout_rate"], list):
        config["dropout_rate"] = config["dropout_rate"][0]
    if isinstance(config["n_ui_layers"], list):
        config["n_ui_layers"] = config["n_ui_layers"][0]
    if isinstance(config["image_knn_k"], list):
        config["image_knn_k"] = config["image_knn_k"][0]
    if isinstance(config["text_knn_k"], list):
        config["text_knn_k"] = config["text_knn_k"][0]
    if isinstance(config["reg_weight"], list):
        config["reg_weight"] = config["reg_weight"][0]

    # Now instantiate model
    model = get_model("SMORE")(config, train_data).to(config["device"])

    return model
````

````python
def load_model_tac_gia(checkpoint_tac_gia):
    # 1. Khởi tạo model trống với kiến trúc hiện tại của bạn
    config = Config("SMORE", "sports", {}, mg=False)
    model = load_model(config)

    # 2. Lấy state_dict từ checkpoint
    ckpt_dict = (
        checkpoint_tac_gia["state_dict"]
        if "state_dict" in checkpoint_tac_gia
        else checkpoint_tac_gia
    )

    # 3. Lọc lấy các lớp trùng tên VÀ trùng kích thước
    model_dict = model.state_dict()
    valid_weights = {}

    for k, v in ckpt_dict.items():
        if k in model_dict:
            if v.shape == model_dict[k].shape:
                valid_weights[k] = v
            else:
                print(
                    f"⚠️ Bỏ qua {k} do lệch size: Checkpoint {v.shape} vs Model {model_dict[k].shape}"
                )
        else:
            print(f"❓ Lớp {k} trong checkpoint không có trong code hiện tại.")

    # 4. Nạp đè vào model
    model_dict.update(valid_weights)
    model.load_state_dict(model_dict)
    print(f"✅ Đã nạp thành công {len(valid_weights)} lớp!")

    return model


model_tac_gia = load_model_tac_gia(checkpoint_tac_gia)
````

````python
import json


def get_recommendations_json(
    model,
    df_user_map,
    df_meta_map,
    user_ids=[0, 1, 2],
    top_k=5,
    filename="results.json",
):
    model.eval()
    all_recommendations = {}

    with torch.no_grad():
        for u_id in user_ids:
            # 1. Dự đoán
            user_tensor = torch.tensor([u_id])
            scores = model.full_sort_predict(user_tensor)
            _, top_items_indices = torch.topk(scores, top_k)
            top_items_list = top_items_indices.flatten().tolist()

            # 2. Lấy thông tin người dùng từ df_user_map
            user_row = df_user_map[df_user_map["userID"] == u_id]

            # Khởi tạo thông tin user mặc định
            user_data = {
                "userID": u_id,
                "reviewerID": "Unknown",
                "reviewerName": "Unknown",
            }

            if not user_row.empty:
                # Trích xuất đúng tên cột từ dataframe gốc của bạn
                user_data["reviewerID"] = user_row.iloc[0]["reviewerID"]
                # Xử lý NaN cho reviewerName
                name = user_row.iloc[0]["reviewerName"]
                user_data["reviewerName"] = name if pd.notna(name) else None

            # 3. Lấy danh sách sản phẩm (giống như các bước trước)
            product_details = []
            for i_id in top_items_list:
                product_row = df_meta_map[df_meta_map["itemID"] == i_id]
                if not product_row.empty:
                    cols_to_keep = ["itemID", "asin", "title", "price", "imUrl"]
                    existing_cols = [
                        c for c in cols_to_keep if c in product_row.columns
                    ]
                    item_dict = product_row[existing_cols].to_dict("records")[0]
                    # Fix NaN cho item
                    item_dict = {
                        k: (v if pd.notna(v) else None) for k, v in item_dict.items()
                    }
                    product_details.append(item_dict)
                else:
                    product_details.append({"itemID": i_id, "error": "Not found"})

            # 4. Gộp vào kết quả với cấu trúc tách biệt
            all_recommendations[str(u_id)] = {
                "reviewerID": user_data["reviewerID"],
                "reviewerName": user_data["reviewerName"],
                "recommendations": product_details,
            }

    # Ghi trực tiếp vào file
    with open(filename, "w", encoding="utf-8") as f:
        json.dump(all_recommendations, f, indent=4, ensure_ascii=False)

    print(f"Đã lưu kết quả vào file: {filename}")
    return all_recommendations


# --- CÁCH SỬ DỤNG ---
# results = get_recommendations_json(model, df_user_map, df_meta_map, filename="amazon_rec.json")
````
