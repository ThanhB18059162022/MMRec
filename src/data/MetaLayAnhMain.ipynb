```python
import os
```

```python
PATH = "F:/"
DIR_NAME = "chunk_3"
```

```python
import tarfile

with tarfile.open(os.path.join(PATH, f"{DIR_NAME}.tar.gz"), "r:gz") as tar:
    tar.extractall(PATH)
```

```python
import os

output_dir = os.path.join(PATH, f'{DIR_NAME}_main_images')
os.makedirs(output_dir, exist_ok=True)
print(f"Directory '{output_dir}' created or already exists.")
```

```python
import shutil
import os

source_dir = os.path.join(PATH, DIR_NAME)

for root, dirs, files in os.walk(source_dir):
    if 'MAIN.jpg' in files:
        # Extract the subdirectory name (e.g., '200000' from 'chunk_4/200000')
        subdir_name = os.path.basename(root)

        # Construct the full path to the source 'MAIN.jpg'
        source_main_jpg_path = os.path.join(root, 'MAIN.jpg')

        # Construct the new filename (e.g., '200000.jpg')
        new_filename = f"{subdir_name}.jpg"

        # Construct the full destination path in the output directory
        destination_path = os.path.join(output_dir, new_filename)

        # Copy and rename the file
        shutil.copyfile(source_main_jpg_path, destination_path)
        print(f"Copied '{source_main_jpg_path}' to '{destination_path}'")
```

```python
import tarfile

input_dir = output_dir
output_archive = f"{output_dir}.tar.gz"

with tarfile.open(output_archive, "w:gz") as tar:
    tar.add(input_dir, arcname=os.path.basename(input_dir))

print(f"Directory '{input_dir}' compressed to '{output_archive}'.")
```
