# Copy các file đã chạy từ các bước bỏ vào thư mục data ngoài để train model


```python
import os
import shutil
```

```python
PATH = "./data/2023"
DATASET_NAME = "baby"
```

```python
file_to_copy = {
    "df_inter.label.inter": f"{DATASET_NAME}.inter",
    "text_feat.npy": "text_feat.npy",
    "image_feat.npy": "image_feat.npy",
    "user_graph_dict.npy": "user_graph_dict.npy"
}
```

```python
# Kiểm tra các file đã tồn tại chưa
for src_file in file_to_copy.keys():
    src_path = os.path.join(PATH, src_file)
    if not os.path.exists(src_path):
        raise FileNotFoundError(f"File {src_file} không tồn tại ở đường dẫn {src_path}. Vui lòng kiểm tra lại.")
    
print("Tất cả file đều đã tồn tại.")
```

```python
# Copy các file đó bỏ vào thư mục data ngoài để train model
for src_file, dst_file in file_to_copy.items():
    src_path = os.path.join(PATH, src_file)
    dst_path = os.path.join(f"../data/{DATASET_NAME}", dst_file)
    os.makedirs(os.path.dirname(dst_path), exist_ok=True)  # Tạo thư mục nếu chưa tồn tại
    shutil.copy(src_path, dst_path)
    print(f"Đã copy {src_file} đến {dst_path}")
```
