```python
# Bỏ giá trị null với trùng
learner_id, course_id, tmstmp_str = 'userID', 'itemID', 'timestamp'

df.dropna(subset=[learner_id, course_id, tmstmp_str], inplace=True)
df.drop_duplicates(subset=[learner_id, course_id, tmstmp_str], inplace=True)
print(f'After dropped: {df.shape}')
df[:3]
```

Trong bài toán GCN và Hệ thống gợi ý, mục đích của code này là loại bỏ những người dùng và sản phẩm có quá ít tương tác để giảm độ nhiễu và giải quyết vấn đề "dữ liệu quá thưa thớt" (Data Sparsity). Bỏ những dòng mà số cặp userID-itemID	< 5.

```python
from collections import Counter
import numpy as np

min_u_num, min_i_num = 5, 5

# Hàm này có nhiệm vụ tìm ra danh sách các ID "không hợp lệ" dựa trên số lượng tương tác.
def get_illegal_ids_by_inter_num(df, field, max_num=None, min_num=None):
    if field is None:
        return set()
    if max_num is None and min_num is None:
        return set()

    max_num = max_num or np.inf
    min_num = min_num or -1

    ids = df[field].values
    inter_num = Counter(ids)
    ids = {id_ for id_ in inter_num if inter_num[id_] < min_num or inter_num[id_] > max_num}
    print(f'{len(ids)} illegal_ids_by_inter_num, field={field}')

    return ids

# Đây là hàm quan trọng nhất, thực hiện lọc dữ liệu theo vòng lặp cho đến khi mọi User và Item đều đạt chuẩn "5-core".
def filter_by_k_core(df):
    while True:
        ban_users = get_illegal_ids_by_inter_num(df, field=learner_id, max_num=None, min_num=min_u_num)
        ban_items = get_illegal_ids_by_inter_num(df, field=course_id, max_num=None, min_num=min_i_num)
        if len(ban_users) == 0 and len(ban_items) == 0:
            return

        dropped_inter = pd.Series(False, index=df.index)
        if learner_id:
            dropped_inter |= df[learner_id].isin(ban_users)
        if course_id:
            dropped_inter |= df[course_id].isin(ban_items)
        print(f'{len(dropped_inter)} dropped interactions')
        df.drop(df.index[dropped_inter], inplace=True)

```

Đoạn code này thực hiện bước Re-indexing (Đánh số lại ID). Đây là một bước "sống còn" sau khi bạn đã lọc dữ liệu (K-core).Nói một cách đơn giản: Nó chuyển các ID loằng ngoằng của Amazon (ví dụ: A3SG769SK) thành các con số nguyên liên tiếp ($0, 1, 2, 3...$) để máy tính có thể xử lý nhanh và dùng làm chỉ mục (index) cho ma trận Embedding.

```python
i_mapping_file = 'i_id_mapping.csv'
u_mapping_file = 'u_id_mapping.csv'

splitting = [0.8, 0.1, 0.1]
uid_field, iid_field = learner_id, course_id

# 1. Khai báo và lấy danh sách ID duy nhất
uni_users = pd.unique(df[uid_field])
uni_items = pd.unique(df[iid_field])

# 2. Tạo "Từ điển" ánh xạ (Mapping Dictionary)
# start from 0
u_id_map = {k: i for i, k in enumerate(uni_users)}
i_id_map = {k: i for i, k in enumerate(uni_items)}

# 3. Cập nhật trực tiếp vào dữ liệu (Mapping)
df[uid_field] = df[uid_field].map(u_id_map)
df[iid_field] = df[iid_field].map(i_id_map)
df[uid_field] = df[uid_field].astype(int)
df[iid_field] = df[iid_field].astype(int)

# 4. Lưu lại "Cuốn sổ địa chỉ" (Mapping File)
# dump
rslt_dir = './'
u_df = pd.DataFrame(list(u_id_map.items()), columns=['user_id', 'userID'])
i_df = pd.DataFrame(list(i_id_map.items()), columns=['asin', 'itemID'])

u_df.to_csv(os.path.join(rslt_dir, u_mapping_file), sep='\t', index=False)
i_df.to_csv(os.path.join(rslt_dir, i_mapping_file), sep='\t', index=False)
print(f'mapping dumped...')
```

Đoạn code này là bước chuẩn bị các "điểm cắt" (split points) để chia tập dữ liệu gốc thành 3 phần: Train (Huấn luyện), Validation (Kiểm tra/Tối ưu) và Test (Thử nghiệm).

Trong nghiên cứu của bạn, việc chia này cực kỳ quan trọng để đảm bảo mô hình SMORE không "học vẹt" mà thực sự có khả năng dự đoán sản phẩm mới.

```python

# =========2. splitting
print(f'splitting ...')
# 1. Chuẩn hóa tỷ lệ (Normalization)
tot_ratio = sum(splitting) # Tổng tỷ lệ (ví dụ: 0.8 + 0.1 + 0.1 = 1.0)
# remove 0.0 in ratios
ratios = [i for i in splitting if i > .0] # Loại bỏ các phần tử bằng 0 (nếu có)
ratios = [_ / tot_ratio for _ in ratios] # Đảm bảo tổng các tỷ lệ luôn bằng 1
# 2. Tạo điểm cắt bằng Tổng tích lũy (Cumulative Sum)
split_ratios = np.cumsum(ratios)[:-1]

#df[tmstmp_str] = df[tmstmp_str].map(lambda x: datetime.strptime(x, "%Y-%m-%dT%H:%M:%SZ"))
split_ratios
```


Tại sao lại là [0.8, 0.9]?
Vì để chia một cái bánh thành 3 phần, bạn chỉ cần 2 nhát cắt:

Nhát 1 tại vị trí 80%: Mọi thứ trước đó là Train.

Nhát 2 tại vị trí 90%: Mọi thứ từ 80% đến 90% là Validation.

Phần còn lại (từ 90% đến 100%) mặc định là Test.

Gemini đã nói
Đoạn code này thực hiện bước Chia dữ liệu theo thời gian (Temporal Splitting).

Trong hệ thống gợi ý, đây là cách chia dữ liệu "chuẩn" nhất vì nó mô phỏng đúng thực tế: Bạn dùng dữ liệu trong quá khứ để huấn luyện mô hình và dùng dữ liệu mới nhất (tương lai) để kiểm tra xem mô hình dự đoán có đúng không.

```python
ts_id = 'timestamp'

# 1. Xác định "Mốc thời gian" để cắt
# Cơ chế: Hàm np.quantile sẽ tìm ra giá trị thời gian (timestamp) tại các điểm 80% và 90% của toàn bộ dữ liệu.
# Ví dụ: Nếu dữ liệu của bạn kéo dài từ năm 2020 đến 2025, nó sẽ tính toán và trả về:
# split_timestamps[0]: Ngày nào đó vào năm 2024 (mốc 80%).
# split_timestamps[1]: Ngày nào đó vào đầu năm 2025 (mốc 90%).

split_timestamps = list(np.quantile(df[ts_id], split_ratios))

# 2. Tách dữ liệu thành 3 tập
# Train (0): Những tương tác xảy ra trước mốc 80%. Đây là dữ liệu mô hình SMORE sẽ "học".
# Validation (1): Những tương tác xảy ra trong khoảng 80% - 90%. Dùng để chỉnh các tham số (Hyperparameters).
# Test (2): Những tương tác mới nhất (10% cuối cùng). Đây là bài thi cuối kỳ để đánh giá độ chính xác của mô hình.
# get df training dataset unique users/items
df_train = df.loc[df[ts_id] < split_timestamps[0]].copy()
df_val = df.loc[(split_timestamps[0] <= df[ts_id]) & (df[ts_id] < split_timestamps[1])].copy()
df_test = df.loc[(split_timestamps[1] <= df[ts_id])].copy()

# 3. Đánh nhãn (Labeling) và Gộp lại
# Thay vì lưu thành 3 file riêng biệt, code này gộp chúng lại thành một bảng duy nhất nhưng thêm một cột "nhãn" (x_label).
# Dòng nào có số 0 -> Thuộc tập Train.
# Dòng nào có số 2 -> Thuộc tập Test.
# Việc này giúp việc quản lý dữ liệu gọn gàng hơn.
# Định dạng .inter: Đây là định dạng phổ biến trong các framework GCN (như RecBole hoặc SMORE).
# Thực chất nó là một file văn bản bình thường nhưng các cột cách nhau bằng dấu Tab (\t).
x_label, rslt_file = 'x_label', 'sports14-indexed.inter'
df_train[x_label] = 0
df_val[x_label] = 1
df_test[x_label] = 2
temp_df = pd.concat([df_train, df_val, df_test])
temp_df = temp_df[[learner_id, course_id, 'rating', ts_id, x_label]]
print(f'columns: {temp_df.columns}')

temp_df.columns = [learner_id, course_id, 'rating', ts_id, x_label]

temp_df.to_csv(os.path.join(rslt_dir, rslt_file), sep='\t', index=False)
temp_df[:5]
#print('done!')
```
