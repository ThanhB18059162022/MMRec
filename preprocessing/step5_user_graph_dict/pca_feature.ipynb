```python
# ===== Load feature =====
text_feat = np.load(
    os.path.join(PATH, "text_feat.npy")
)
image_feat = np.load(
    os.path.join(PATH,  "image_feat.npy")
)
print("Text shape:", text_feat.shape)
print("Image shape:", image_feat.shape)
```

```python
# ===== Normalize (RẤT QUAN TRỌNG) =====
text_feat = text_feat / (np.linalg.norm(text_feat, axis=1, keepdims=True) + 1e-8)
image_feat = image_feat / (np.linalg.norm(image_feat, axis=1, keepdims=True) + 1e-8)
```

```python
# Lưa giá trị normalize
np.save(
    os.path.join(PATH, "text_feat.normalize.npy"),
    text_feat,
)

np.save(
    os.path.join(PATH, "image_feat.normalize.npy"),
    image_feat,
)
```

```python
# ===== PCA =====
text_pca_dim = 384
image_pca_dim = 384
```

```python
print("Fitting PCA for image...")
pca_img = PCA(n_components=image_pca_dim)
image_feat_pca = pca_img.fit_transform(image_feat)
```

```python
print("Fitting PCA for text...")
pca_txt = PCA(n_components=text_pca_dim)
text_feat_pca = pca_txt.fit_transform(text_feat)
```

```python
# ===== Normalize lại sau PCA (khuyến nghị) =====
image_feat_pca = image_feat_pca / (
    np.linalg.norm(image_feat_pca, axis=1, keepdims=True) + 1e-8
)
text_feat_pca = text_feat_pca / (
    np.linalg.norm(text_feat_pca, axis=1, keepdims=True) + 1e-8
)
```

```python
print("Image explained variance:", np.sum(pca_img.explained_variance_ratio_))
print("Text explained variance:", np.sum(pca_txt.explained_variance_ratio_))
```

```python
# ===== Save =====
np.save(os.path.join(PATH, f"image_feat.{image_pca_dim}.npy"), image_feat_pca)
np.save(
    os.path.join(PATH, f"text_feat.{text_pca_dim}.npy"),
    text_feat_pca,
)

print("Done!")
print("Image shape:", image_feat_pca.shape)
print("Text shape:", text_feat_pca.shape)
```
