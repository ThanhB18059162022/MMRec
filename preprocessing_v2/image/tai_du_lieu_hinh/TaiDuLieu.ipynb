```python
def download_url(dir_name):
  download_url = f"https://mcauleylab.ucsd.edu/public_datasets/data/amazon_2023/raw/review_categories/{dir_name}"

  if "meta" in dir_name:
    download_url = f"https://mcauleylab.ucsd.edu/public_datasets/data/amazon_2023/raw/meta_categories/{dir_name}"

  output_filename = os.path.join(PATH, dir_name)
  !wget -O "{output_filename}" "{download_url}"
```

```python
dir_name = "meta_Video_Games.jsonl.gz"

# download_url(dir_name)
```

```python
import gzip

dir_name_gz = "meta_Video_Games.jsonl.gz"
input_filepath_gz = os.path.join(PATH, dir_name_gz)

dir_name_jsonl = dir_name_gz.replace(".gz", "")
output_filepath_jsonl = os.path.join(PATH, dir_name_jsonl)

with gzip.open(input_filepath_gz, 'rb') as f_in:
    with open(output_filepath_jsonl, 'wb') as f_out:
        f_out.write(f_in.read())

print(f"Successfully unzipped {dir_name_gz} to {dir_name_jsonl}")
```
