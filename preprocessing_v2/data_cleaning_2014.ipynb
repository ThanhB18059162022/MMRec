```python
import os
import pandas as pd
```

```python
df_meta = pd.read_parquet(os.path.join(PATH, "df_meta_5core_2014.parquet"))
```

```python
df_meta_to_drop = df_meta[df_meta["imUrl"].isna()].copy()
df_meta_to_drop.shape
```

```python
## Ghi ra lại file
df_meta.to_parquet(os.path.join(PATH, "clean/df_meta.parquet"), index=False)
```

```python
df_inter_map_i_id = df_inter.merge(df_i_id_map, how="inner", on="itemID")
```
