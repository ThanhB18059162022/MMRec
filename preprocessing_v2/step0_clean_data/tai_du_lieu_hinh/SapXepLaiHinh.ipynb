```python
reviews_chunks = pd.read_json(os.path.join(PATH, "2023/Baby_Products.jsonl"), lines=True, chunksize=100_000)
```

```python
# Get unique parent_asin values from df_meta_2023 for efficient lookup
meta_parent_asins = set(df_meta_2023['parent_asin'].unique())

# Initialize lists to store chunks
all_reviews_in_meta_chunks = []
all_reviews_not_in_meta_chunks = []

print("Starting full chunk processing...")
# Iterate through all chunks
for i, chunk in enumerate(reviews_chunks):
    print(f"Processing chunk {i+1}...")
    # Create a boolean mask
    is_in_meta_mask = chunk['parent_asin'].isin(meta_parent_asins)

    # Append filtered chunks to respective lists
    all_reviews_in_meta_chunks.append(chunk[is_in_meta_mask].copy())
    all_reviews_not_in_meta_chunks.append(chunk[~is_in_meta_mask].copy())

# Concatenate all collected chunks into final DataFrames
df_reviews_in_meta = pd.concat(all_reviews_in_meta_chunks, ignore_index=True)
df_reviews_not_in_meta = pd.concat(all_reviews_not_in_meta_chunks, ignore_index=True)

print("\n--- First Part: Reviews Processing Complete ---")
print("DataFrame with reviews whose parent_asin is in df_meta_2023:")
display(df_reviews_in_meta.head())
print(f"Shape of df_reviews_in_meta: {df_reviews_in_meta.shape}")

print("\nDataFrame with reviews whose parent_asin is NOT in df_meta_2023:")
display(df_reviews_not_in_meta.head())
print(f"Shape of df_reviews_not_in_meta: {df_reviews_not_in_meta.shape}")

# --- Second Part: Meta-data not in Reviews ---

# Get all unique parent_asin values from the processed reviews (both in_meta and not_in_meta)
all_reviews_parent_asins = pd.concat([
    df_reviews_in_meta['parent_asin'],
    df_reviews_not_in_meta['parent_asin']
]).unique()

# Convert to a set for efficient lookup
set_all_reviews_parent_asins = set(all_reviews_parent_asins)

# Find parent_asin values in df_meta_2023 that are NOT in the reviews
mask_meta_not_in_reviews = ~df_meta_2023['parent_asin'].isin(set_all_reviews_parent_asins)
df_meta_not_in_reviews = df_meta_2023[mask_meta_not_in_reviews].copy()

print("\n--- Second Part: Meta-data Processing Complete ---")
print("Meta-data whose parent_asin is NOT in any of the processed reviews:")
display(df_meta_not_in_reviews.head())
print(f"Shape of df_meta_not_in_reviews: {df_meta_not_in_reviews.shape}")
```

```python
import os
import re

def get_numerical_first_and_last_id(directory_path):
    """Gets the first and last image IDs (numerically) from a directory."""
    if not os.path.exists(directory_path):
        return f"Directory '{directory_path}' does not exist.", None, None

    # List files and filter for .jpg images, extracting the numerical part
    image_ids = []
    for f in os.listdir(directory_path):
        if f.endswith('.jpg'):
            match = re.match(r'(\d+)\.jpg', f)
            if match:
                image_ids.append(int(match.group(1)))

    if not image_ids:
        return f"No .jpg images found in '{directory_path}'.", None, None

    min_id = min(image_ids)
    max_id = max(image_ids)
    total_count = len(image_ids)

    return min_id, max_id, total_count

# Define the directories to check again
directories_to_check = {
    "chunk_0_main_images": "/content/chunk_0_main_images",
    "chunk_1_main_images": "/content/chunk_1_main_images",
    "chunk_2_main_images": "/content/chunk_2_main_images",
    "chunk_3_main_images": "/content/chunk_3_main_images",
    "chunk_4_main_images": "/content/chunk_4_main_images",
}

print("Re-evaluating image ID ranges and counts:")
for name, path in directories_to_check.items():
    min_id, max_id, count = get_numerical_first_and_last_id(path)
    if min_id is None: # Error or empty directory message
        print(f"For {name}: {path} - {min_id}")
    else:
        print(f"For {name} (Directory: {path}):\n  Lowest Image ID: {min_id}\n  Highest Image ID: {max_id}\n  Total .jpg files: {count}\n")
```

```python
import os
import re
import shutil
import pandas as pd

# 1. Create df_index to parent_asin mapping
# Ensure df_index is set if it's not already the index (it was set previously)
id_to_asin_map = df_meta_2023.set_index('df_index')['parent_asin'].to_dict()

# Define old and new chunk directories
current_chunk_dirs = {
    "chunk_0_main_images": "/content/chunk_0_main_images",
    "chunk_1_main_images": "/content/chunk_1_main_images",
    "chunk_2_main_images": "/content/chunk_2_main_images",
    "chunk_3_main_images": "/content/chunk_3_main_images",
    "chunk_4_main_images": "/content/chunk_4_main_images",
}

new_base_dir = "/content/" # New chunks will be directly in /content/

# Mapping from old directory base name to new directory base name
old_to_new_dir_names = {
    "chunk_0_main_images": "chunk_0_images",
    "chunk_1_main_images": "chunk_1_images",
    "chunk_2_main_images": "chunk_2_images",
    "chunk_3_main_images": "chunk_3_images",
    "chunk_4_main_images": "chunk_4_images",
}

print("\nStarting image reorganization and renaming...")
moved_count = 0
skipped_count = 0
collision_count = 0 # Track actual collisions during move

for old_dir_key, old_dir_path in current_chunk_dirs.items():
    if not os.path.exists(old_dir_path):
        print(f"Source directory {old_dir_path} does not exist. Skipping.")
        continue

    new_dir_name = old_to_new_dir_names.get(old_dir_key)
    if not new_dir_name:
        print(f"Could not determine new directory name for {old_dir_key}. Skipping.")
        continue
    new_dir_path = os.path.join(new_base_dir, new_dir_name)

    os.makedirs(new_dir_path, exist_ok=True)
    print(f"Processing images from '{old_dir_path}' to '{new_dir_path}'...")

    for filename in os.listdir(old_dir_path):
        if not filename.endswith('.jpg'):
            continue

        match = re.match(r'(\d+)\.jpg', filename)
        if not match:
            print(f"Skipping non-standard image filename: {filename} in {old_dir_path}")
            skipped_count += 1
            continue

        original_df_index = int(match.group(1))
        source_file_path = os.path.join(old_dir_path, filename)

        if original_df_index in id_to_asin_map:
            new_asin = id_to_asin_map[original_df_index]
            if pd.isna(new_asin): # Check for NaN parent_asin
                print(f"Skipping {filename}: parent_asin is NaN for df_index {original_df_index}.")
                skipped_count += 1
                continue
            new_filename = f"{new_asin}.jpg"
            destination_file_path = os.path.join(new_dir_path, new_filename)

            if os.path.exists(destination_file_path):
                print(f"Collision detected: {new_filename} already exists in {new_dir_path}. Overwriting.")
                collision_count += 1
            shutil.move(source_file_path, destination_file_path)
            moved_count += 1
        else:
            print(f"Skipping {filename}: df_index {original_df_index} not found in df_meta_2023 map.")
            skipped_count += 1

print("\nImage reorganization complete.")
print(f"Total images moved and renamed: {moved_count}")
print(f"Total images skipped (invalid ID or NaN ASIN): {skipped_count}")
print(f"Total detected filename collisions (overwrites): {collision_count}")

# Optionally, remove the old directories if they are empty
print("\nAttempting to remove old, now empty, chunk directories...")
for old_dir_key, old_dir_path in current_chunk_dirs.items():
    if os.path.exists(old_dir_path) and not os.listdir(old_dir_path):
        os.rmdir(old_dir_path)
        print(f"Removed empty directory: {old_dir_path}")
    elif os.path.exists(old_dir_path):
        print(f"Directory {old_dir_path} is not empty or does not exist. Not removing.")
```

```python
import os
import random
import pandas as pd
import json
import shutil

# Define paths
base_image_dir = '/content/'
chunk_dirs = [
    'chunk_0_images',
    'chunk_1_images',
    'chunk_2_images',
    'chunk_3_images',
    'chunk_4_images'
]
output_sample_dir = '/content/random_image_sample'

# Create the output directory if it doesn't exist
os.makedirs(output_sample_dir, exist_ok=True)

print(f"Created output directory: {output_sample_dir}")

sampled_items = []

for chunk_dir_name in chunk_dirs:
    full_chunk_path = os.path.join(base_image_dir, chunk_dir_name)

    if not os.path.exists(full_chunk_path):
        print(f"Warning: Chunk directory '{full_chunk_path}' does not exist. Skipping.")
        continue

    # Get a list of all .jpg files in the current chunk directory
    image_files = [f for f in os.listdir(full_chunk_path) if f.endswith('.jpg')]

    if not image_files:
        print(f"Warning: No .jpg images found in '{full_chunk_path}'. Skipping.")
        continue

    # Select a random image file
    random_image_filename = random.choice(image_files)
    random_image_path = os.path.join(full_chunk_path, random_image_filename)
#
    # Extract the parent_asin from the filename (e.g., 'B01C4319LO.jpg' -> 'B01C4319LO')
    parent_asin = os.path.splitext(random_image_filename)[0]

    # Find the corresponding row in df_meta_2023
    row_data = df_meta_2023[df_meta_2023['parent_asin'] == parent_asin]

    if not row_data.empty:
        # Save the row data as a JSON file
        json_output_path = os.path.join(output_sample_dir, f'{parent_asin}.json')
        row_data.to_json(json_output_path, orient='records', indent=4)
        print(f"Saved JSON for ASIN '{parent_asin}' to '{json_output_path}'.")

        # Copy the image file to the output directory
        shutil.copy(random_image_path, output_sample_dir)
        print(f"Copied image '{random_image_filename}' to '{output_sample_dir}'.")

        sampled_items.append({'chunk': chunk_dir_name, 'parent_asin': parent_asin, 'image_file': random_image_filename})
    else:
        print(f"Warning: No matching metadata found for ASIN '{parent_asin}' (image: {random_image_filename}). Skipping JSON and image copy.")

print("\n--- Sampling Summary ---")
if sampled_items:
    for item in sampled_items:
        print(f"From {item['chunk']}: ASIN = {item['parent_asin']}, Image = {item['image_file']}")
else:
    print("No items were sampled.")
```

```python
import os
import shutil
import pandas as pd

# Create a new directory for filtered images
output_filtered_dir = '/content/filtered_2014_images'
os.makedirs(output_filtered_dir, exist_ok=True)
print(f"Created output directory: {output_filtered_dir}")

base_image_dir = '/content/'

# Define the df_index ranges for each chunk to map df_index to the correct source directory
# These ranges were determined previously based on the original chunking of df_index
chunk_ranges = {
    'chunk_0_images': (0, 49999),
    'chunk_1_images': (50000, 99999),
    'chunk_2_images': (100000, 149999),
    'chunk_3_images': (150000, 193988),
    'chunk_4_images': (200000, 217723)
}

# --- Optimization: Cache existing image filenames in each chunk directory ---
existing_images_cache = {}
print("Caching existing image filenames...")
for chunk_name in chunk_ranges.keys():
    full_chunk_path = os.path.join(base_image_dir, chunk_name)
    if os.path.exists(full_chunk_path):
        # Store a set of filenames (e.g., 'B01C4319LO.jpg') for quick lookup
        existing_images_cache[chunk_name] = set(f for f in os.listdir(full_chunk_path) if f.endswith('.jpg'))
        print(f"  Cached {len(existing_images_cache[chunk_name])} images from {chunk_name}")
    else:
        existing_images_cache[chunk_name] = set() # No images if directory doesn't exist
        print(f"  Warning: Chunk directory '{full_chunk_path}' does not exist. No images cached.")
print("Finished caching.")
# -------------------------------------------------------------------------

copied_count = 0
not_found_count = 0

print(f"\nProcessing {len(df_meta_2014_filter)} items from df_meta_2014_filter...")

# Iterate through the filtered DataFrame
for index, row in df_meta_2014_filter.iterrows():
    parent_asin = str(row['parent_asin']) # Ensure ASIN is string for filename
    df_index = row['df_index'] # Use the df_index to determine the chunk

    source_chunk_dir_name = None # To store 'chunk_0_images', etc.
    # Determine which chunk directory the image belongs to based on its original df_index
    for chunk_name, (min_idx, max_idx) in chunk_ranges.items():
        if min_idx <= df_index <= max_idx:
            source_chunk_dir_name = chunk_name
            break

    if source_chunk_dir_name:
        image_filename = f"{parent_asin}.jpg"

        # Use the cache for fast existence check
        if image_filename in existing_images_cache.get(source_chunk_dir_name, set()):
            source_image_path = os.path.join(base_image_dir, source_chunk_dir_name, image_filename)
            destination_image_path = os.path.join(output_filtered_dir, image_filename)
            shutil.copy(source_image_path, destination_image_path)
            copied_count += 1
        else:
            not_found_count += 1
            # Print a warning only if the image was expected but not found in cache
            print(f"Warning: Image '{image_filename}' not found in expected directory '{os.path.join(base_image_dir, source_chunk_dir_name)}' for df_index {df_index}.")
    else:
        not_found_count += 1
        print(f"Warning: Could not determine source chunk directory for df_index {df_index} (ASIN: {parent_asin}).")

print(f"\n--- Image Copying Summary ---")
print(f"Total images copied: {copied_count}")
print(f"Total images not found or chunk directory not determined: {not_found_count}")
```

```python
missing_image_asins = []

print("Re-checking for missing images...")

for index, row in df_meta_2014_filter.iterrows():
    parent_asin = str(row['parent_asin'])
    df_index = row['df_index']

    source_chunk_dir_name = None
    for chunk_name, (min_idx, max_idx) in chunk_ranges.items():
        if min_idx <= df_index <= max_idx:
            source_chunk_dir_name = chunk_name
            break

    if source_chunk_dir_name:
        image_filename = f"{parent_asin}.jpg"
        if image_filename not in existing_images_cache.get(source_chunk_dir_name, set()):
            missing_image_asins.append({
                'parent_asin': parent_asin,
                'df_index': df_index,
                'expected_chunk_dir': os.path.join(base_image_dir, source_chunk_dir_name)
            })
    else:
        missing_image_asins.append({
            'parent_asin': parent_asin,
            'df_index': df_index,
            'expected_chunk_dir': 'Unknown'
        })

print(f"Found {len(missing_image_asins)} missing images.")
print("Details of missing images:")
for item in missing_image_asins:
    print(f"  ASIN: {item['parent_asin']}, df_index: {item['df_index']}, Expected Path: {item['expected_chunk_dir']}")
```

```python
print("Checking df_index values for items with 'Unknown' expected paths:")
unknown_path_df_indices = []
for item in missing_image_asins:
    if item['expected_chunk_dir'] == 'Unknown':
        unknown_path_df_indices.append(item['df_index'])

if unknown_path_df_indices:
    print(f"Found {len(unknown_path_df_indices)} df_index values with 'Unknown' paths: {sorted(unknown_path_df_indices)}")

    # Define the gap based on previous chunk_ranges
    gap_start = chunk_ranges['chunk_3_images'][1] + 1  # Max_idx of chunk 3 + 1
    gap_end = chunk_ranges['chunk_4_images'][0] - 1  # Min_idx of chunk 4 - 1
    print(f"The identified gap in df_index ranges is from {gap_start} to {gap_end}.")

    in_gap_count = 0
    for idx in unknown_path_df_indices:
        if gap_start <= idx <= gap_end:
            print(f"  df_index {idx} (ASIN: {df_meta_2023.loc[idx, 'parent_asin']}) falls within the gap.")
            in_gap_count += 1
        else:
            print(f"  df_index {idx} (ASIN: {df_meta_2023.loc[idx, 'parent_asin']}) does NOT fall within the gap, further investigation needed.")
    print(f"Total df_index values falling within the gap: {in_gap_count}")

else:
    print("No df_index values with 'Unknown' expected paths found to analyze.")
```

```python
import tarfile
import os

# Define the base directory where the new chunk_X_images directories are located
base_directory = '/content/'

# List of all new chunk directories to compress
new_chunk_dirs = [
    'chunk_0_images',
    'chunk_1_images',
    'chunk_2_images',
    'chunk_3_images',
    'chunk_4_images',
]

print("Starting compression of all reorganized image chunks...")
for chunk_dir_name in new_chunk_dirs:
    source_directory_path = os.path.join(base_directory, chunk_dir_name)
    output_filename = os.path.join(base_directory, f'{chunk_dir_name}.tar.gz')

    if os.path.exists(source_directory_path):
        print(f"Compressing directory '{source_directory_path}' to '{output_filename}'...")
        try:
            with tarfile.open(output_filename, 'w:gz') as tar:
                # Add the entire directory. The arcname ensures the archive contains
                # the files directly under the chunk_dir_name, not nested path.
                tar.add(source_directory_path, arcname=os.path.basename(source_directory_path))
            print(f"Successfully compressed '{source_directory_path}'.")
        except Exception as e:
            print(f"Error compressing {source_directory_path}: {e}")
    else:
        print(f"Source directory '{source_directory_path}' does not exist. Skipping.")

print("\nAll specified chunk directories have been processed for compression.")
```
